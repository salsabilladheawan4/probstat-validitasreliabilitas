```python
import re
import pandas as pd
import numpy as np
from scipy import stats
```

```python
nama_file = "/content/ujidata.xlsx"
print("File terbaca:", nama_file)
```

```python
JUMLAH_KOLOM_DEMOGRAFI = 6   # file ini punya 6 kolom non-butir

df = pd.read_excel(nama_file)
butir = df.iloc[:, JUMLAH_KOLOM_DEMOGRAFI:]
butir = butir.loc[:, butir.columns.str.contains(r"\[")]   # buang kolom Durasi

print("Jumlah responden :", len(df))
print("Jumlah butir     :", butir.shape[1])
print("Data kosong      :", int(butir.isna().sum().sum()))
```

```python
skala = {
    "Sangat Tidak Setuju (STS)": 1,
    "Tidak Setuju (TS)": 2,
    "Netral (N)": 3,
    "Setuju (S)": 4,
    "Sangat Setuju (SS)": 5,
}
num = butir.replace(skala).apply(pd.to_numeric, errors="coerce")
print("Sel yang gagal dikonversi:", int(num.isna().sum().sum()))
num.head()
```

```python
dimensi = {}
for kolom in butir.columns:
    nama = re.match(r"^(.*?) \[", kolom).group(1).strip()
    dimensi.setdefault(nama, []).append(kolom)

for nama, daftar in dimensi.items():
    print(f"{nama}: {len(daftar)} butir")
```

```python
n = len(num)
t = stats.t.ppf(0.975, n - 2)
r_tabel = t / np.sqrt((n - 2) + t**2)
print(f"n = {n} | r tabel (α=0,05; df={n-2}) = {r_tabel:.3f}")

hasil_validitas = []
for nama, daftar in dimensi.items():
    x = num[daftar]
    total = x.sum(axis=1)
    for i, kolom in enumerate(daftar, 1):
        r, p = stats.pearsonr(x[kolom], total)
        r_koreksi, _ = stats.pearsonr(x[kolom], total - x[kolom])
        hasil_validitas.append({
            "Dimensi": nama,
            "Butir": f"B{i}",
            "r hitung": round(r, 3),
            "Corrected r": round(r_koreksi, 3),
            "p-value": round(p, 4),
            "Keterangan": "Valid" if r > r_tabel else "Tidak valid",
        })

tabel_validitas = pd.DataFrame(hasil_validitas)
tabel_validitas
```

```python
def cronbach_alpha(x):
    k = x.shape[1]
    return k / (k - 1) * (1 - x.var(ddof=1).sum() / x.sum(axis=1).var(ddof=1))

def kategori(a):
    return ("Sangat baik" if a >= 0.90 else "Baik" if a >= 0.80
            else "Reliabel" if a >= 0.70 else "Tidak reliabel")

hasil_reliabilitas, alpha_dihapus = [], []
for nama, daftar in dimensi.items():
    x = num[daftar]
    a = cronbach_alpha(x)
    hasil_reliabilitas.append({"Dimensi": nama, "Jumlah butir": len(daftar),
                               "Cronbach's alpha": round(a, 3), "Kategori": kategori(a)})
    for i, kolom in enumerate(daftar, 1):
        if len(daftar) > 2:
            alpha_dihapus.append({"Dimensi": nama, "Butir": f"B{i}",
                "Alpha jika dihapus": round(cronbach_alpha(x.drop(columns=kolom)), 3)})

tabel_reliabilitas = pd.DataFrame(hasil_reliabilitas)
tabel_alpha_dihapus = pd.DataFrame(alpha_dihapus)
display(tabel_reliabilitas)
display(tabel_alpha_dihapus)
```

```python
gabung = tabel_validitas.merge(tabel_alpha_dihapus, on=["Dimensi", "Butir"], how="left")
with pd.ExcelWriter("hasil_uji_validitas_reliabilitas.xlsx") as w:
    gabung.to_excel(w, sheet_name="Validitas", index=False)
    tabel_reliabilitas.to_excel(w, sheet_name="Reliabilitas", index=False)
print("Tersimpan di panel Files")
```

```python
nama_dim = list(dimensi.keys())[0]            # dimensi Penggunaan AI
kolom_buang = dimensi[nama_dim][3]            # B4
print("Butir dibuang:", kolom_buang)

alpha_sebelum = cronbach_alpha(num[dimensi[nama_dim]])

dimensi_baru = {k: list(v) for k, v in dimensi.items()}
dimensi_baru[nama_dim].remove(kolom_buang)

x = num[dimensi_baru[nama_dim]]
total = x.sum(axis=1)
alpha_sesudah = cronbach_alpha(x)
print(f"Alpha sebelum: {alpha_sebelum:.3f} | sesudah: {alpha_sesudah:.3f}")

for i, kolom in enumerate(dimensi_baru[nama_dim], 1):
    r, p = stats.pearsonr(x[kolom], total)
    print(f"B{i}: r hitung = {r:.3f} -> {'Valid' if r > r_tabel else 'Tidak valid'}")
```

```python
sampel = num.sample(20, random_state=1)
n2 = len(sampel)
t2 = stats.t.ppf(0.975, n2 - 2)
r_tabel2 = t2 / np.sqrt((n2 - 2) + t2**2)
print(f"n = {n2} | r tabel = {r_tabel2:.3f}")

jumlah_valid, jumlah_butir = 0, 0
for nama, daftar in dimensi.items():
    xs = sampel[daftar]
    tot = xs.sum(axis=1)
    for kolom in daftar:
        r, _ = stats.pearsonr(xs[kolom], tot)
        jumlah_butir += 1
        jumlah_valid += int(r > r_tabel2)

print(f"Butir valid: {jumlah_valid} dari {jumlah_butir}")
```

```python
def kategori_mean(m):
    return ("Sangat rendah" if m <= 1.80 else "Rendah" if m <= 2.60
            else "Sedang" if m <= 3.40 else "Tinggi" if m <= 4.20 else "Sangat tinggi")

rows = []
for nama, daftar in dimensi.items():
    m = num[daftar].values.mean()
    rows.append({"Dimensi": nama, "Rata-rata": round(m, 2), "Kategori": kategori_mean(m)})
pd.DataFrame(rows)
```
